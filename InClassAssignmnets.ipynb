```python
def checkEmailAddress(email):
    try:
        assert isinstance(email, str), "Input must be a string"

        parts = email.split("@")
        if len(parts) != 2:
            raise ValueError("Email must contain exactly one @")

        username = parts[0]
        domain_full = parts[1]

        domain_parts = domain_full.split(".")
        if len(domain_parts) < 2:
            raise ValueError("Domain part must contain at least one dot")

        domain = domain_parts[0]
        topmain = ".".join(domain_parts[1:])

        if username == "" or domain == "" or topmain == "":
            raise ValueError("Username, domain, and topmain must not be empty")

        return True

    except (AssertionError, ValueError):
        return False

checkEmailAddress("example@domain.com")
```

```python
def getValues (n):
  assert isinstance(n, int), "'n' must be an integer"
  items = []

  for i in range(n):
    items.append (float(input("Enter a value:")))

  if len(items) < 2:
    raise ValueError("the length of list 'items' cannot be < 2")
  return items


def calcAverage(alist):
  assert isinstance(alist, list), "'alist' must be a list"
  try:
    return sum(alist) / len(alist)
  except Exception as e:
    print(f"Error: {e}")


def main():
  n = int(input("'Enter a number of items to be added to the list: "))
  alist = getValues (n)
  avg = calcAverage(alist)
  print (f"Average weight: {avg} lbs.")
main ()
```
