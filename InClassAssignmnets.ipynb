```python
Sales = int(input("Enter the total sales: "))
print(f"Total Sales: {Sales}")
COGS = 4800
GP = Sales - COGS
print(f"Gross Profit: {GP}")
TaxRate = float(input("Enter Tax Rate: "))
print(f"Tax Rate: {TaxRate}")
print(f"Cost of Goods Sold: {COGS}")
Tax = TaxRate * GP
print(f"Total Taxes: {Tax}")
NI = GP - Tax
print(f"Net Income: {NI}")
```

In Class Assignment #2

```python
x = float(input("Enter the first number: "))
y = float(input("Enter the second number: "))
op = input("Enter the operator: ")

if op == "+":
  answer = x + y
elif op == "-":
  answer = x - y
elif op == "*":
  answer = x * y
elif op == "/":
  answer = x / y
elif op == "//":
  answer = x // y
elif op == "**":
  answer = x ** y
elif op == "%":
  answer = x % y
else:
  print("Invalid Operator")
print(answer)
```

In Class Assignmnet #3

```python
flatfee = 5
hourly = 2.5
print("Hours|Total Fee")

for hours in range(1,9,1):
  fee = flatfee + (hours*hourly)
  if fee > 20:
    print(f"   {hours} | 20")
  elif fee < 10:
    print(f"   {hours} | 10")
  else:
    print(f"   {hours} | {fee}")
```

In Class Assignmnet #4

```python
ticket = 10
print(f"Ad$ | Profit")

for ads in range(0,201,25):
  addattended = (2 * (ads**.5))
  totalattended = addattended + 20
  revenue = totalattended * ticket
  costs = ads + 200
  profit = round(revenue - costs)

  print(f" {ads} | {profit}")

```

In Class Assignment #5

```python
flatfee = 5
hourly = 2.5
print("Hours|Total Fee")

hours = 1
while hours < 9:
  fee = flatfee + (hours*hourly)
  if fee > 20:
    print(f"   {hours} | 20")
  elif fee < 10:
    print(f"   {hours} | 10")
  else:
    print(f"   {hours} | {fee}")
  hours += 1
```

In Class Assignment #6

```python
import random

play = input("Do you want to play? (y/n): ")

while play == 'y':
  play = input("Choose rock, paper, or scissors: ").lower()
  computer = random.choice(['rock','paper','scissors'])
  print(f"Computer chose {computer}")

  if player == computer:
    print("Draw")

  elif (
      player == 'rock' and computer == 'scissors' or
      player == 'paper' and computer == 'rock' or
      player == 'scissors' and computer == 'paper'
  ):
    print("You win")
  else:
    print("You lose")

  play = input("Do you want to continue playing? (y/n): ")
```

```python
numbers = []

while True:
    user = float(input("Enter a number (enter 0 to stop): "))
    if user == 0:
        break
    numbers.append(user)

numbers.sort(reverse=True)
print("Numbers in descending order:", numbers)
```
