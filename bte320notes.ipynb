```python
grade = float(input("Enter student grade: "))
if grade >= 97:
  print("A+")
elif grade >= 93:
  print("A")
elif grade >= 90:
  print("A-")
elif grade >= 87:
  print("B+")
elif grade >= 83:
  print("B")
elif grade >= 80:
  print("B-")
elif grade >= 77:
  print("C+")
elif grade >= 73:
  print("C")
elif grade >= 70:
  print("C-")
elif grade >= 67:
  print("D+")
elif grade >= 63:
  print("D")
elif grade >= 60:
  print("D-")
else:
  print("F")
```

```python
for TaxRate in range(20,50,10):
  Sales = float(input("Enter the total sales: "))
  print(f"Total Sales: {Sales}")
  COGS = 4800
  GP = Sales - COGS
  print(f"Gross Profit: {GP}")
  TaxRate = float(input("Enter Tax Rate: "))
  print(f"Tax Rate: {TaxRate}")
  print(f"Cost of Goods Sold: {COGS}")
  Tax = TaxRate * GP
  print(f"Total Taxes: {Tax}")
  NI = GP - Tax
  print(f"Net Income: {NI}")
```
