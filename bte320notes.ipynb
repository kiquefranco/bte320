```python
def quickstart(L):
  if len(L) < 2:
    return L
  else:
    pivot = L(len(L)//2)
    less, equal, larger = [], [], []

    for item in L:
      if item < pivot:
        less.append(item)
      elif item == pivot:
        equal.append(item)
      else:
        larger.append(item)

  return quickstart(less) + equal + quickstart(larger)
quickstart([0,2,5,1,6,3,7,4])
```

```python
class Car:
  def __init__(self, model, make, year, type, engine):
      self.model = model
      self.make = make
      self.year = year
      self.car_type = type
      self.engine = engine
      self.num_wheels = 4
      self.num_doors = 5
      self.engine_running = False
      self.speed = 0

  def engine_Start(self):
      self.engine_running = True
      print("Engine Started!")

  def engine_stop(self):
      if not self.engine_running:
        print("The Engine is already stopped")
      else:
        self.engine_running = False
        print("Engine Stopped")

  def accelerate(self, new_speed):
      self.speed = new_speed
      if self.speed > 150:
        print("Cannot go any faster")
        self.speed -= new_speed

  def decelerate(self, new_speed):
      self.speed -= new_speed
      if self.speed < 0:
        print("Cannot go any slower")
        self.speed += new_speed

  def __str__(self):
      return f"Make: {self.make}; Model: {self.model}"
```
